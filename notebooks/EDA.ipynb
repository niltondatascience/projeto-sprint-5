## Histograma — quilometragem

Distribuição da coluna `odometer` para entender o perfil de uso dos veículos anunciados.

```python
fig = px.histogram(
    car_data,
    x='odometer',
    nbins=50,
    title='Distribuição da quilometragem dos veículos',
    labels={'odometer': 'Quilometragem (milhas)'},
)
fig.show()
```

## Histograma — preço por tipo de veículo

```python
fig = px.histogram(
    car_data,
    x='price',
    color='type',
    nbins=50,
    title='Distribuição de preços por tipo de veículo',
    labels={'price': 'Preço (US$)', 'type': 'Tipo'},
)
fig.show()
```

## Gráfico de dispersão — quilometragem x preço

Verificação da relação entre a quilometragem e o preço anunciado, separada pela condição do veículo.

```python
fig = px.scatter(
    car_data,
    x='odometer',
    y='price',
    color='condition',
    opacity=0.5,
    title='Relação entre quilometragem e preço',
    labels={
        'odometer': 'Quilometragem (milhas)',
        'price': 'Preço (US$)',
        'condition': 'Condição',
    },
)
fig.show()
```

## Gráfico de dispersão — ano do modelo x preço

```python
fig = px.scatter(
    car_data,
    x='model_year',
    y='price',
    color='type',
    opacity=0.5,
    title='Relação entre ano do modelo e preço',
    labels={
        'model_year': 'Ano do modelo',
        'price': 'Preço (US$)',
        'type': 'Tipo',
    },
)
fig.show()
```

## Conclusões

- A maior parte dos anúncios se concentra em veículos com quilometragem entre 50 mil e 200 mil milhas.
- Existe relação inversa entre quilometragem e preço: quanto maior a quilometragem, menor o valor anunciado.
- Veículos mais recentes tendem a apresentar preços mais altos, com grande dispersão entre os tipos.
- As colunas `model_year`, `cylinders`, `odometer`, `paint_color` e `is_4wd` apresentam valores ausentes.
